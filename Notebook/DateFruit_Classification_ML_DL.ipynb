## 11. Create Empty List for storing Model Results : 

```python
results = []
```

## 12. Expected Calibration Error Function : 

```python
def expected_calibration_error(y_true, y_prob, n_bins = 10) :
    bins = np.linspace(0, 1, n_bins+1)
    binids = np.digitize(y_prob,bins) - 1
    ece = 0

    for i in range(n_bins) :
        mask = binids==i

        if np.sum(mask) > 0 :
            accuracy = np.mean(y_true[mask])
            confidence = np.mean(y_prob[mask])
            ece += np.abs(accuracy-confidence) * np.sum(mask) / len(y_true)

    return ece
```

## 13. Evaluation Function : 

```python
def evaluate_model(model_name, y_test, y_pred, y_prob) :
    accuracy = accuracy_score(y_test,y_pred)
    precision = precision_score(y_test, y_pred, average = "weighted")
    recall = recall_score(y_test, y_pred, average = "weighted")
    f1 = f1_score(y_test, y_pred, average = "weighted")
    cm = confusion_matrix(y_test,y_pred)
    sensitivity = np.mean(np.diag(cm)/cm.sum(axis=1))
    specificity = []

    for i in range(len(cm)) :
        TP = cm[i, i]
        FN = cm[i, :].sum()-TP
        FP = cm[:, i].sum()-TP
        TN = cm.sum()-TP-FN-FP
        specificity.append(TN/(TN+FP))

    specificity = np.mean(specificity)
    y_test_bin = label_binarize(y_test, classes = np.unique(y_test))

    roc_auc = roc_auc_score(y_test_bin, y_prob, average = "weighted", multi_class = "ovr")
    brier = np.mean([brier_score_loss(y_test_bin[:,i], y_prob[:, i])
        for i in range(y_prob.shape[1])
    ])
    ece = np.mean([
        expected_calibration_error(y_test_bin[:, i], y_prob[:, i])
        for i in range(y_prob.shape[1])
    ])
    results.append({
        "Model" : model_name,
        "Accuracy" : accuracy,
        "Precision" : precision,
        "Recall" : recall,
        "F1 Score" : f1,
        "ROC AUC" : roc_auc,
        "Sensitivity" : sensitivity,
        "Specificity" : specificity,
        "Brier Score" : brier,
        "ECE" : ece
    })
    return cm
```

## 14. Confusion Matrix Function : 

```python
def plot_confusion_matrix(cm, title) :
    plt.figure(figsize = (7,6))

    sns.heatmap(cm, annot = True, cmap = "Blues", fmt = "d")

    plt.title(title)
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    
    plt.show()
```

## 15. ROC Curve Function : 

```python
def plot_multiclass_roc(y_test, y_prob, title) :
    y_test_bin = label_binarize(y_test, classes = np.unique(y_test))

    plt.figure(figsize = (8, 6))

    for i in range(y_prob.shape[1]) :
        fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_prob[:, i])
        plt.plot(fpr, tpr, lw = 2, label = f"Class {i}")

    plt.plot([0, 1], [0, 1], "--", color = "black")
    plt.title(title)
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")

    plt.legend()
    plt.grid(True)
    plt.show()
```

## 16. Make Leaderboard : 

```python
leaderboard = pd.DataFrame(results)

if leaderboard.empty :
    print("No model results found! Train and evaluate models first.")
else :

    # Sort by Accuracy
    leaderboard = leaderboard.sort_values(
        by="Accuracy",
        ascending=False
    ).reset_index(drop=True)

    # Medal Icons
    medals = ["🥇", "🥈", "🥉"]

    rank = []

    for i in range(len(leaderboard)):
        if i < 3:
            rank.append(medals[i])
        else:
            rank.append("🏅")

    leaderboard.insert(0, "Rank", rank)

    # Round values
    leaderboard = leaderboard.round(4)

    # Display Beautiful Table
    display(
        leaderboard.style
        .background_gradient(cmap="Blues", subset=["Accuracy"])
        .background_gradient(cmap="Purples", subset=["Precision"])
        .background_gradient(cmap="Greens", subset=["Recall"])
        .background_gradient(cmap="Oranges", subset=["F1 Score"])
        .background_gradient(cmap="RdPu", subset=["ROC AUC"])
        .background_gradient(cmap="YlGn", subset=["Sensitivity"])
        .background_gradient(cmap="BuPu", subset=["Specificity"])
        .background_gradient(cmap="Reds_r", subset=["Brier Score"])
        .background_gradient(cmap="Reds_r", subset=["ECE"])
        .set_caption("🏆 Final Model Leaderboard")
        .hide(axis="index")
    )
```

## 72. Create Clustering Leaderboard : 

```python
cluster_results = []
```

## 73. Evaluation Function : 

```python
def evaluate_clustering(model_name, X, labels) :
    unique_labels = np.unique(labels)
    n_clusters = len(unique_labels)

    if -1 in unique_labels :
        n_clusters -= 1

    if n_clusters < 2 :
        print(f"{model_name}: Less than 2 clusters detected.")
        return

    silhouette = silhouette_score(X, labels)
    calinski = calinski_harabasz_score(X, labels)
    davies = davies_bouldin_score(X, labels)
    cluster_results.append({
        "Model": model_name,
        "Clusters": n_clusters,
        "Silhouette Score": silhouette,
        "Calinski-Harabasz": calinski,
        "Davies-Bouldin": davies
    })

    print(f"{model_name} Evaluation Completed.")
```

## 119. Device Configeration : 

```python
import torch 

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using Device : ", device)
```

## 120. Train Test Split : 

```python
X_train_dl, X_test_dl, y_train_dl, y_test_dl = train_test_split(X_scaled, y, test_size = 0.2, random_state = 42, stratify = y)

print("Training Shape : ", X_train_dl.shape)
print("Testing Shape  : ", X_test_dl.shape)
```

```python
print(type(X_train_dl))
print(type(y_train_dl))
```

## 121. Convert Data To Tensors : 

```python
X_train_tensor = torch.FloatTensor(X_train_dl.to_numpy())
X_test_tensor = torch.FloatTensor(X_test_dl.to_numpy())


y_train_tensor = torch.LongTensor(y_train_dl.to_numpy())
y_test_tensor = torch.LongTensor(y_test_dl.to_numpy())

print("Training Tensor Shape : ", X_train_tensor.shape)
print("Testing Tensor Shape  : ", X_test_tensor.shape)
print("Training Label Shape  : ", y_train_tensor.shape)
print("Testing Label Shape   : ", y_test_tensor.shape)
```

## 122. Create Tensor Dataset and DataLoader : 

```python
from torch.utils.data import TensorDataset, DataLoader

train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
test_dataset = TensorDataset(X_test_tensor, y_test_tensor)

BATCH_SIZE = 64

train_loader = DataLoader(train_dataset, batch_size = BATCH_SIZE, shuffle = True)
test_loader = DataLoader(test_dataset, batch_size = BATCH_SIZE, shuffle = False)

print("Train Batches : ", len(train_loader))
print("Test Batches  : ", len(test_loader))
```

## 123. Feed Forward Neural Network Model : 

```python
import torch.nn as nn

class FNN(nn.Module) :
    def __init__(self, input_size, num_classes) :
        super(FNN, self).__init__()

        self.network = nn.Sequential(
            nn.Linear(input_size, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(0.30),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(0.30),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, num_classes)
        )

    def forward(self, x) :
        return self.network(x)


import torch.optim as optim 

INPUT_SIZE = X_train_tensor.shape[1]
NUM_CLASSES = len(torch.unique(y_train_tensor))

fnn_model = FNN(input_size = INPUT_SIZE, num_classes=NUM_CLASSES).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(fnn_model.parameters(), lr = 0.001, weight_decay = 1e-4)

print(fnn_model)
print("Loss Function : ", criterion)
print("Optimizer : ", optimizer)
```

## 124. Hyperparameters : 

```python
EPOCHS = 100

train_losses = []
test_losses = []

train_accuracies = []
test_accuracies = []

best_accuracy = 0.0

print("Epochs : ", EPOCHS)
```

## 125. Training Loop : 

```python
import copy
import time

best_model_wts = copy.deepcopy(fnn_model.state_dict())
best_accuracy = 0.0
start_time = time.time()

for epoch in range(EPOCHS) :
    fnn_model.train()
    train_loss = 0.0
    train_correct = 0
    train_total = 0

    for inputs, labels in train_loader :
        inputs = inputs.to(device)
        labels = labels.to(device)
        optimizer.zero_grad()
        outputs = fnn_model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        train_loss += loss.item() * inputs.size(0)
        _, predicted = torch.max(outputs, 1)
        train_correct += (predicted == labels).sum().item()
        train_total += labels.size(0)

    epoch_train_loss = train_loss / train_total
    epoch_train_acc = train_correct / train_total

    train_losses.append(epoch_train_loss)
    train_accuracies.append(epoch_train_acc)


    fnn_model.eval()
    test_loss = 0.0
    test_correct = 0
    test_total = 0

    with torch.no_grad() :
        for inputs, labels in test_loader :
            inputs = inputs.to(device)
            labels = labels.to(device)
            outputs = fnn_model(inputs)
            loss = criterion(outputs, labels)
            test_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            test_correct += (predicted == labels).sum().item()
            test_total += labels.size(0)

    epoch_test_loss = test_loss / test_total
    epoch_test_acc = test_correct / test_total

    test_losses.append(epoch_test_loss)
    test_accuracies.append(epoch_test_acc)

    if epoch_test_acc > best_accuracy :
        best_accuracy = epoch_test_acc
        best_model_wts = copy.deepcopy(fnn_model.state_dict())

    print(f"Epoch [{epoch+1:03d}/{EPOCHS}] | "f"Train Loss: {epoch_train_loss:.4f} | "f"Train Acc: {epoch_train_acc:.4f} | "f"Test Loss: {epoch_test_loss:.4f} | "f"Test Acc: {epoch_test_acc:.4f}")


training_time = time.time() - start_time

print("\nTraining Finished Successfully!")
print(f"Best Test Accuracy : {best_accuracy:.4f}")
print(f"Training Time : {training_time:.2f} seconds")
```
